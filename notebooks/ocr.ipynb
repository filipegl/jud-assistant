```python
from ocr.file_handler import extract_text
from dotenv import load_dotenv

load_dotenv()
OCR_LANG = os.getenv("OCR_LANG")
TESSDATA_DIR = os.getenv("TESSDATA_DIR")
```

This `extract_text` function will convert the pdf pages into images and then use the ocr module to extract text.

```python
pdf_path = "docs/acordao_examples/habeascorpus.pdf"
text = extract_text(pdf_path, lang=OCR_LANG, tessdata_dir=TESSDATA_DIR)
```
